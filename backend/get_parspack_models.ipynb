# دریافت مدل‌های ParsPack
این نوت‌بوک `AI_GATEWAY_BASE_URL` و `AI_GATEWAY_API_KEY` را از فایل `.env` پروژه می‌خواند.

```python
import os
from html import escape
from pathlib import Path

import aiohttp
from IPython.display import HTML, display

env_file = next((path for path in (Path(".env"), Path("backend/.env")) if path.exists()), None)
assert env_file, "backend/.env was not found"
env = dict(line.strip().split("=", 1) for line in env_file.read_text(encoding="utf-8").splitlines() if line.strip() and not line.lstrip().startswith("#") and "=" in line)
base_url = os.getenv("AI_GATEWAY_BASE_URL") or env.get("AI_GATEWAY_BASE_URL", "https://ai.parspack.com/")
api_key = os.getenv("AI_GATEWAY_API_KEY") or env.get("AI_GATEWAY_API_KEY")
assert api_key, "AI_GATEWAY_API_KEY is not configured"
url = f"{base_url.rstrip('/')}/v1/models"
headers = {"Authorization": f"Bearer {api_key}"}

async with aiohttp.ClientSession(headers=headers) as session:
    async with session.get(url, timeout=20) as response:
        response.raise_for_status()
        payload = await response.json(content_type=None)

models = sorted(item["id"] for item in payload.get("data", []) if item.get("id"))
assert models, "ParsPack returned no models"
Path("parspack_models.txt").write_text("\n".join(models), encoding="utf-8")
print(f"{len(models)} models — full list saved to parspack_models.txt")
display(HTML(f"<pre style='max-height:600px;overflow:auto'>{escape(chr(10).join(models))}</pre>"))
```
